**What to remember from this assignment:**

1. Representationl power is a function of the number of trainable
   parameters of a model.
2. A simple logistic regression model performs a linear discrimination, it
   is only capable of creating a linear decision boundary in the decision
   space to separate the categories it is trying to predict.
3. Binary classification network models are a sequence of `Dense` layers
   that use a nonlinear activation function like `relu` as the output
   of the layers.
4. Binary classification models should have final `Dense` layer with
   a single unit that uses `sigmoid` activation, to output a prediction
   in the range 0-1
5. Binary classification models should usually use `binary_crossentropy`
   for their loss function, and the default `rmsprop` optimizer is usually
   fine for these types of problems.
6. A model that is incapable of solving a problem will usually show
   no ability to be trained to get performance better than a baseline
   random guessing metric.
8. A model without enough representationl power will have validation loss
   curves that keep getting smaller, there will be no sign that validation
   loss starts getting worse if the model is underpowered.
